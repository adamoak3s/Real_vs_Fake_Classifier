## Library Imports

```python
import os, shutil, pathlib, math, kagglehub
from os import listdir
```

# Download Data from Kaggle

## CIFAKE: Real and AI-Generated Synthetic Images

### https://www.kaggle.com/datasets/birdy654/cifake-real-and-ai-generated-synthetic-images

#### Dataset details

The dataset contains two classes - REAL and FAKE.

For REAL, we collected the images from Krizhevsky & Hinton's CIFAR-10 dataset

For the FAKE images, we generated the equivalent of CIFAR-10 with Stable Diffusion version 1.4

There are 100,000 images for training (50k per class) and 20,000 for testing (10k per class)

```python
# Download latest version
path = kagglehub.dataset_download("birdy654/cifake-real-and-ai-generated-synthetic-images")

print("Path to dataset files:", path)
```

# Split the data (before runtime)

#### Test Data is already split. I need to split the training set into train & validation. I will use an 80/20 split.  Will populate ./images directory like this:
    images
        train
            REAL
            FAKE
        validation
            REAL
            FAKE
        test
            REAL
            FAKE

```python
original_base_dir = pathlib.Path(path)
new_base_dir = pathlib.Path("../data")

#split the training data into training and validation
for label in ["REAL", "FAKE"]:
    original_dir = original_base_dir / "train" / label

    #get the list of files
    files = [f for f in listdir(original_dir)]
    count = len(files)

    #copy over training
    os.makedirs(new_base_dir / "train" / label, exist_ok=True)
    train_end_index = math.floor(count * .8)
    for fname in files[:train_end_index]:
        shutil.copyfile(src=original_dir / fname, dst=new_base_dir / "train" / label / fname)

    #copy over validation
    os.makedirs(new_base_dir / "validation" / label, exist_ok=True)
    for fname in files[train_end_index:]:
        shutil.copyfile(src=original_dir / fname, dst=new_base_dir / "validation" / label / fname)

#copy over the testing data
shutil.copytree(src=original_base_dir / "test", dst=new_base_dir / "test", dirs_exist_ok=True)  # Fine
```
